```python
merged = orders.merge(order_items, on='order_id')
merged
```

```python
merged['total_price'] = merged['quantity'] * merged['list_price'] * (1 - merged['discount'])
merged['total_price'] 
```

```python
reference_date = merged['order_date'].max() + dt.timedelta(days=1)
reference_date
```

```python
import datetime as dt
```

```python
rfm = merged.groupby('customer_id').agg(
    recency=('order_date', lambda x: (reference_date - x.max()).days),
    frequency=('order_id', 'nunique'),
    monetary=('total_price', 'sum')
).reset_index()
rfm
```

```python
rfm.describe()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
scaler = StandardScaler()
rfm_scaled = scaler.fit_transform(rfm[['recency', 'frequency', 'monetary']])
```

```python
scaler
rfm_scaled
```

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
rfm['cluster'] = kmeans.fit_predict(rfm_scaled)
```

```python
rfm['cluster'].value_counts()
```

```python
cluster_summary = rfm.groupby('cluster')[['recency', 'frequency', 'monetary']].mean().round(2)
cluster_summary
```

```python
segment_map = {
    0: 'At Risk',
    1: 'Medium Value',
    2: 'High Value'
}
```

```python
segment_map
```

```python
rfm['segment'] = rfm['cluster'].map(segment_map)
rfm[['customer_id', 'recency', 'frequency', 'monetary', 'segment']].head(10)
```

```python
final_segments = rfm[['customer_id', 'recency', 'frequency', 'monetary', 'segment']]
final_segments
```

```python
final_segments.to_sql('customer_segments', con=engine, if_exists='replace', index=False)
```

# Chart 1 — Number of Customers by Segment

```python
plt.figure(figsize=(8, 5))
plt.bar(segment_counts.index, segment_counts.values,color=({'green','red','purple'}))
plt.title('Number of Customers by Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Number of Customers')
plt.grid(color='black')
plt.show()
```

# Chart 2 — Average Revenue by Segment

```python
avg_monetary = rfm.groupby('segment')['monetary'].mean()
avg_monetary
```

```python
plt.figure(figsize=(8, 5))
plt.bar(avg_monetary.index, avg_monetary.values,color=({'brown','pink','grey'}))
plt.title('Average Revenue per Customer by Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Average Revenue')
plt.grid(color='black')
plt.show()
```
